```python
import os
import time
```

```python
folder_name = input("Enter folder name: ")

if os.path.exists(folder_name):
    os.rmdir(folder_name)
    print("Folder deleted.")
else:
    os.mkdir(folder_name)
    print("Folder created.")
```

```python
path = input("Enter the main path: ")

folders = input("Enter folder names separated by space: ").split()

for folder in folders:
    folder_path = os.path.join(path, folder)
    os.mkdir(folder_path)

print("Folders created.")
```

```python
file_path = input("Enter file path: ")

file_size = os.path.getsize(file_path)
file_size_mb = file_size / (1024 * 1024)

print(f"File size: {file_size_mb:.2f} MB")
```

```python
path = input("Enter folder path: ")

old_files = set(os.listdir(path))

while True:
    time.sleep(5)

    new_files = set(os.listdir(path))

    added_files = new_files - old_files
    deleted_files = old_files - new_files

    for file in added_files:
        print(f"New file added: {file}")

    for file in deleted_files:
        print(f"File deleted: {file}")

    old_files = new_files
```

```python
path = input("Enter folder path: ")

image_extensions = [".jpg", ".jpeg", ".png", ".gif", ".webp", ".bmp", ".tiff", ".tif", ".svg", ".ico"]
audio_extensions = [".mp3", ".wav", ".flac", ".aac", ".ogg", ".m4a"]
video_extensions = [".mp4", ".avi", ".mkv", ".mov", ".wmv", ".webm", ".flv"]
document_extensions = [".pdf", ".doc", ".docx", ".txt", ".rtf", ".odt"]

for file in os.listdir(path):

    file_path = os.path.join(path, file)

    if os.path.isfile(file_path):

        extension = os.path.splitext(file)[1].lower()

        if extension in image_extensions:
            folder = "image"

        elif extension in audio_extensions:
            folder = "audio"

        elif extension in video_extensions:
            folder = "video"

        elif extension in document_extensions:
            folder = "document"

        else:
            continue

        folder_path = os.path.join(path, folder)

        if not os.path.exists(folder_path):
            os.mkdir(folder_path)

        new_path = os.path.join(folder_path, file)

        os.rename(file_path, new_path)

print("Files sorted.")
```
